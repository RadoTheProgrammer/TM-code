```python
import sys


def binary_to_ascii(text):
    """Decode whitespace-separated 8-bit binary values as ASCII."""
    values = text.split()
    characters = []
    for value in values:
        if len(value) != 8 or any(bit not in "01" for bit in value):
            raise ValueError(f"Invalid 8-bit binary value: {value!r}")
        characters.append(chr(int(value, 2)))
    return "".join(characters)


if __name__ == "__main__":
    try:
        with open("ascii-chaussettes.txt") as f:
            print(binary_to_ascii(f.read()))
    except ValueError as error:
        print(error, file=sys.stderr)
        sys.exit(1)
```
